```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    cohen_kappa_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    roc_curve,
)

sns.set(style="whitegrid")

# Output Directories
OUTPUT_DIR = "OUTPUT"
FIG_DIR = os.path.join(OUTPUT_DIR, "figures")
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(FIG_DIR, exist_ok=True)

# Helper function to save and show plots
def save_and_show(fig, filename: str):
    path = os.path.join(FIG_DIR, filename)
    fig.tight_layout()
    fig.savefig(path, bbox_inches="tight")
    print("Saved figure:", path)
    plt.show()

#evaluates any classification models
def evaluate_model(y_true, y_pred, y_proba=None, model_name="Model"):
    acc = accuracy_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    Cohen_kappa = cohen_kappa_score(y_true, y_pred)
    auc = roc_auc_score(y_true, y_proba) if y_proba is not None else None

    print(f"\n=== {model_name} Evaluation ===")
    print(f"Accuracy       : {acc:.4f}")
    print(f"F1 Score       : {f1:.4f}")
    print(f"Cohen's Kappa  : {Cohen_kappa:.4f}")
    if auc is not None:
        print(f"ROC AUC        : {auc:.4f}")
    print("\nClassification Report:\n")
    print(classification_report(y_true, y_pred))

    return {"accuracy": acc, "f1": f1, "Cohen_kappa": Cohen_kappa, "auc": auc}

```

```python
# Load Dataset
df = pd.read_csv(r"DATASET2 NEW\DATASET2_NEW.csv")
print("Original shape:", df.shape)
display(df.head())
df.info()

# Handle Missing Values
# Text columns
df['description'] = df['description'].fillna('')
df['location'] = df['location'].fillna('')
df['lang'] = df['lang'].fillna('unknown')

# Image URLs / paths
image_cols = ['profile_image_url', 'profile_background_image_url', 'profile_image_path', 'profile_background_image_path']
for col in image_cols:
    df[col] = df[col].fillna('no_image')

# Numeric columns
numeric_cols = ['favourites_count','followers_count','friends_count','statuses_count','average_tweets_per_day','account_age_days','split']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(df[col].median())

# Boolean columns
bool_cols = ['default_profile','default_profile_image','geo_enabled','verified']
for col in bool_cols:
    df[col] = df[col].astype(int)

# Encode target
df['account_type'] = df['account_type'].map({'human':0,'bot':1})

```

```python
df.info()
```

```python
# Target Visualization
fig, ax = plt.subplots(figsize=(6,4))
sns.countplot(x=df['account_type'], palette=['tab:green','tab:red'], ax=ax)
ax.set_title("Bot vs Human Distribution")
ax.set_xlabel("Label (0=Human, 1=Bot)")
ax.set_ylabel("Count")
save_and_show(fig, "target_distribution.png")
```

```python
# Encode Categorical Features
le_lang = LabelEncoder()
df['lang_encoded'] = le_lang.fit_transform(df['lang'])

# Feature Engineering
df['followers_friends_ratio'] = df['followers_count'] / (df['friends_count'] + 1)
df['tweets_per_day'] = df['statuses_count'] / (df['account_age_days'] + 1)
df['favourites_per_day'] = df['favourites_count'] / (df['account_age_days'] + 1)

# Description-based features
df['description_length'] = df['description'].apply(len)
df['description_has_url'] = df['description'].apply(lambda x: 1 if 'http' in x else 0)
df['description_has_mentions'] = df['description'].apply(lambda x: 1 if '@' in x else 0)
df['description_has_hashtag'] = df['description'].apply(lambda x: 1 if '#' in x else 0)

# Numeric Feature Distributions 
feature_cols = [
    'followers_count','friends_count','statuses_count','favourites_count',
    'followers_friends_ratio','tweets_per_day','favourites_per_day',
    'description_length','description_has_url','description_has_mentions','description_has_hashtag',
    'default_profile','default_profile_image','geo_enabled','verified','lang_encoded','account_age_days'
]

colors = ['tab:blue','tab:orange','tab:green','tab:red','tab:purple','tab:brown']

for i, col in enumerate(feature_cols):
    fig, ax = plt.subplots(figsize=(6,4))
    sns.histplot(df[col], bins=50, kde=True, color=colors[i % len(colors)], ax=ax)
    ax.set_title(f"Distribution of {col}")
    ax.set_xlabel(col)
    save_and_show(fig, f"{col}_distribution.png")

```

```python
# Correlation Heatmap
fig, ax = plt.subplots(figsize=(12,8))
corr = df[feature_cols + ['account_type']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", ax=ax)
ax.set_title("Feature Correlation Heatmap")
save_and_show(fig, "correlation_heatmap.png")

```

```python
# Prepare Feature Matrix
X = df[feature_cols]
y = df['account_type']

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```python
# Train/Test Split 
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)
print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)
print("Train class balance:\n", y_train.value_counts(normalize=True))

```

```python
#train model - Random Forest classifier
rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
y_proba_rf = rf.predict_proba(X_test)[:,1]
```

```python
#train model - Gradient Boosting classifier
gb = GradientBoostingClassifier(n_estimators=200, learning_rate=0.1, random_state=42)
gb.fit(X_train, y_train)
y_pred_gb = gb.predict(X_test)
y_proba_gb = gb.predict_proba(X_test)[:,1]
```

```python
#evaluate both models using the common helper function
metrics_rf = evaluate_model(y_test, y_pred_rf, y_proba_rf, "Random Forest")
metrics_gb = evaluate_model(y_test, y_pred_gb, y_proba_gb, "Gradient Boosting")

#visualise confusion matrices for both models side by side
fig, axes = plt.subplots(1, 2, figsize=(10,4))

cm_rf = confusion_matrix(y_test, y_pred_rf)
sns.heatmap(cm_rf, annot=True, fmt="d", cmap="Purples", cbar=False, ax=axes[0])
axes[0].set_title("Random Forest – Confusion Matrix")
axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("True")

cm_gb = confusion_matrix(y_test, y_pred_gb)
sns.heatmap(cm_gb, annot=True, fmt="d", cmap="Greens", cbar=False, ax=axes[1])
axes[1].set_title("Gradient Boosting – Confusion Matrix")
axes[1].set_xlabel("Predicted")
axes[1].set_ylabel("True")

save_and_show(fig, "confusion_matrices.png")

#plot ROC curves for both models on the same figure to compare AUC
fig, ax = plt.subplots(figsize=(6,5))

fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf)
fpr_gb, tpr_gb, _ = roc_curve(y_test, y_proba_gb)

auc_rf = metrics_rf["auc"] if metrics_rf["auc"] is not None else 0.5
auc_gb = metrics_gb["auc"] if metrics_gb["auc"] is not None else 0.5

ax.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.3f})", color="tab:purple")
ax.plot(fpr_gb, tpr_gb, label=f"Gradient Boosting (AUC = {auc_gb:.3f})", color="tab:orange")
ax.plot([0,1],[0,1], linestyle="--", color="grey", alpha=0.6)

ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curves")
ax.legend(loc="lower right")
ax.grid(alpha=0.3)

save_and_show(fig, "roc_curves.png")

#simple bar chart comparing Accuracy, F1, and Kappa for both models
metrics_df = pd.DataFrame({
    "model": ["Random Forest", "Random Forest", "Random Forest",
              "Gradient Boosting", "Gradient Boosting", "Gradient Boosting"],
    "metric": ["accuracy", "f1", "Cohen_kappa", "accuracy", "f1", " Cohen_kappa"],
    "value": [
        metrics_rf["accuracy"], metrics_rf["f1"], metrics_rf["Cohen_kappa"],
        metrics_gb["accuracy"], metrics_gb["f1"], metrics_gb["Cohen_kappa"],
    ],
})

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(
    data=metrics_df,
    x="metric",
    y="value",
    hue="model",
    palette=["tab:purple", "tab:orange"],
    ax=ax
)
ax.set_ylim(0, 1.05)
ax.set_title("Model Comparison: Accuracy, F1, Cohen_Kappa")
ax.set_xlabel("")
ax.set_ylabel("Score")

for p in ax.patches:
    height = p.get_height()
    ax.text(
        p.get_x() + p.get_width() / 2,
        height + 0.01,
        f"{height:.2f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

save_and_show(fig, "model_comparison_rf_vs_gb.png")
```

```python
#plot feature importance from the Random Forest model
feat_importances = pd.Series(rf.feature_importances_, index=feature_cols).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x=feat_importances.values, y=feat_importances.index, palette="viridis", ax=ax)
ax.set_title("Random Forest Feature Importance")
ax.set_xlabel("Importance")
ax.set_ylabel("Feature")
save_and_show(fig, "rf_feature_importance.png")

#save the scaled features along with the label to a processed CSV
df_processed = pd.DataFrame(X_scaled, columns=feature_cols)
df_processed['Bot Label'] = y.values

processed_path = os.path.join(OUTPUT_DIR, "twitter_processed_scaled.csv")
df_processed.to_csv(processed_path, index=False)
print("Saved processed dataset to:", processed_path)
```
